### Energy Price Calculator

This code calculates the energy price based on the number of units consumed and the customer type (residential or commercial). It uses a tiered pricing model with `if-else` statements.

```python
print("--- Energy Price Calculator ---")

# Get units from the user
units_input = input("Enter the number of units consumed: ")
units = float(units_input) # Convert input to a floating-point number

# Get customer type from the user
customer_type = input("Enter customer type (residential/commercial/industrial): ")

# Initialize price variable
price = 0.0

# Calculate price using if-else statements based on customer type
if customer_type.lower() == "residential":
    price = units * 2
    print(f"For residential customers, the price is: ${price:.2f}")
elif customer_type.lower() == "commercial":
    price = units * 5
    print(f"For commercial customers, the price is: ${price:.2f}")
elif customer_type.lower() == "industrial":
    price = units * 10
    print(f"For industrial customers, the price is: ${price:.2f}")
else:
    print("Invalid customer type entered. Please enter 'residential', 'commercial', or 'industrial'.")

print("------------------------------")
```
